```python
import torch
```

```python
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
e = torch.exp(x)
e.retain_grad()
s = torch.sum(e, 0)
s.retain_grad()
soft = e / s
soft.retain_grad()
o = torch.sum(soft, 0)
o.backward()
dsoft = torch.tensor([1.0, 1.0, 1.0])
ds = (-e / s**2 * dsoft).sum(0)
de = 1 / s * dsoft
de += torch.full_like(e, ds.item())
dx = e * de
print(torch.allclose(dsoft, soft.grad))
print(torch.allclose(ds, s.grad))
print(torch.allclose(de, e.grad))
print(torch.allclose(dx, x.grad, atol=1e-4))
print(dx)
print(x.grad)
```
