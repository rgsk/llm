```python
import torch
```

```python
rahul = {
    "2019": torch.tensor([70, 80, 90]),
    "2020": torch.tensor([71, 81, 91]),
}
mehak = {
    "2019": torch.tensor([75, 85, 95]),
    "2020": torch.tensor([76, 86, 96]),
}
marks = torch.stack(
    [
        rahul["2019"],
        rahul["2020"],
        mehak["2019"],
        mehak["2020"],
    ]
)
print(marks.shape)
for row in marks:
    print(row)
```

```python
marks = marks.reshape(2, 2, 3)
marks
```

```python
from itertools import permutations

for p in permutations([0, 1, 2]):
    print("-" * 8)
    print(p)
    perm = marks.permute(*p)
    print(perm.shape)
    print(perm)
```
