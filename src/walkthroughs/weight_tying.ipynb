```python
import torch
```

over n_layer blocks the variance would grow ~linearly with the number of adds

```python
times = 1000
x = torch.zeros(100)
for _ in range(times):
    d = torch.randn_like(x)
    x += d
x.var()
```

```python
times = 100
x = torch.zeros(100)
for _ in range(times):
    d = torch.randn_like(x)
    x += d
x.var()
```

Shrinking each contributing projection keeps it flat.

```python
times = 1000
x = torch.zeros(100)
for _ in range(times):
    d = torch.randn_like(x) / times ** 0.5
    x += d
x.var()
```

```python
times = 100
x = torch.zeros(100)
for _ in range(times):
    d = torch.randn_like(x) / times ** 0.5
    x += d
x.var()
```

```python
x = torch.randn(100)
print(x.std())
print((x * 10).std())
print((x / 10).std())
```
