```python
import torch
```

```python
fan_in = 400
res = []
x = torch.randn(fan_in)
for _ in range(10000):
    w = torch.randn(fan_in)
    res.append((x * w).sum().item())
torch.tensor(res).var()
```

```python
fan_in = 400
res = []
x = torch.randn(fan_in)
for _ in range(10000):
    w = torch.randn(fan_in) / fan_in**0.5
    res.append((x * w).sum().item())
torch.tensor(res).var()
```
