```python
import torch
```

```python
fan_in = 400
res = []
for _ in range(10000):
    res.append((torch.randn(fan_in)).sum().item())
torch.tensor(res).var()
```

```python
fan_in = 400
res = []
for _ in range(10000):
    res.append((torch.randn(fan_in) / fan_in ** 0.5).sum().item())
torch.tensor(res).var()
```

```python
fan_in = 400
res = []
x = torch.randn(fan_in)
for _ in range(10000):
    w = torch.randn(fan_in)
    res.append((x * w).sum().item())
torch.tensor(res).var()
```

```python
fan_in = 400
res = []
x = torch.randn(fan_in)
for _ in range(10000):
    w = torch.randn(fan_in) / fan_in**0.5
    res.append((x * w).sum().item())
torch.tensor(res).var()
```

```python
torch.manual_seed(0)
fan_in = 400

# --- Linear: output is a SUM over fan_in terms ---
x = torch.randn(10000, fan_in)              # unit-variance input
W_bad = torch.randn(fan_in, 8)              # N(0, 1)
W_good = torch.randn(fan_in, 8) / fan_in**0.5

print("input std      ", x.std().item())
print("out std N(0,1) ", (x @ W_bad).std().item())    # ~20 = sqrt(400)
print("out std scaled ", (x @ W_good).std().item())   # ~1
```
